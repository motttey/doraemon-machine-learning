```python
API_TOKEN = os.environ["sakura_api_secret"]
```

```python
import requests
# import json
```

```python
def print_chatgpt_response(prompt=''):
	API_KEY = API_TOKEN

	# promptがない場合
	if not prompt:
		return

	api_base = "https://api.ai.sakura.ad.jp/v1/chat/completions"
	headers = {
		"Content-Type": "application/json",
		"Authorization": f"Bearer {API_KEY}",
	}
	model = "gpt-oss-120b"  # モデル
	messages = [{"role": "user", "content": prompt}]
	data = {"model": model, "messages": messages}

	response = requests.post(api_base, headers=headers, json=data)
	print(response.json()["choices"][0]["message"]["content"])
```

```python
prompt = 'あなたはドラえもんです。自己紹介をしてください。'
print_chatgpt_response(prompt)
```
